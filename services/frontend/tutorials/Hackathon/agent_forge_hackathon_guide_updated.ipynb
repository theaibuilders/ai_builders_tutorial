
# Agent Forge AI Hackathon Guide

## Setting up your development environment

### 1. Create a project directory

Open your terminal (Command Prompt or PowerShell on Windows, Terminal on macOS or Linux) and create a new folder for your project:

```bash
mkdir agent-forge-project
cd agent-forge-project
```

Keep all of your files inside this folder. You can name it whatever you like.

### 2. Install Python

You need Python 3. Install it using the method appropriate for your operating system:

- **Linux (Ubuntu/Debian):** Use `apt` to install a recent Python version. First update the package index, then install Python:
  
  ```bash
  sudo apt-get update
  sudo apt-get install python3.6  # or replace 3.6 with a newer version available on your system
  ```
  
  This command installs Python 3 and its dependencies. You can check the version with:
  
  ```bash
  python3 --version
  ```

- **macOS:** The recommended way is to download the official installer from Python.org. Go to the [Python Downloads page](https://www.python.org/downloads/macos/) and click the link for the latest Python 3 release. After the installer downloads, double‑click it and follow the on‑screen instructions to complete the installation. Once finished, open a Terminal and verify the installation:
  
  ```bash
  python3 --version
  ```

- **Windows:** Use the Microsoft Store or the Windows Package Manager. Open PowerShell and run:
  
  ```powershell
  winget install Python.Python.3.14
  ````
  
  Alternatively, search for “Python” in the Microsoft Store and install it. After installation, close and reopen PowerShell or Command Prompt and verify the version:
  
  ```powershell
  python --version
  ```

### 3. Set up a virtual environment

A virtual environment keeps your project’s dependencies isolated from other projects. To create one, run the following command in your project folder (replace `venv` with whatever name you prefer):

```bash
python3 -m venv venv
```

This command creates a directory named `venv` containing a private copy of Python and `pip`.  You can then **activate** the virtual environment. Activation scripts vary by platform:

| Platform      | Command to activate the virtual environment |
|---------------|--------------------------------------------|
| **Linux/macOS bash/zsh** | `source venv/bin/activate` |
| **Linux/macOS fish shell** | `source venv/bin/activate.fish` |
| **Linux/macOS csh/tcsh** | `source venv/bin/activate.csh` |
| **Windows (cmd.exe)** | `venv\Scriptsctivate.bat` |
| **Windows (PowerShell)** | `venv\Scripts\Activate.ps1` |

When the environment is activated, your terminal prompt is prefaced with `(venv)` to indicate you are working in that environment. To deactivate, simply run `deactivate`.

### 4. Install required packages

Once the virtual environment is active, you can install the Python libraries needed for your project using `pip`. For example, to install the `requests` library:

```bash
pip install requests
```

If your project has multiple dependencies, create a `requirements.txt` file listing each package and run:

```bash
pip install -r requirements.txt
```

Installing packages this way ensures that everyone on your team uses the same library versions.



## Organizing your project

A clean folder structure makes your code easier to navigate. Here is a suggested layout for a hackathon project:

```
agent-forge-project/
├── notebooks/            # Jupyter notebooks and experiments
│   └── guide.ipynb      # This tutorial notebook
├── src/                 # Source code for your agent (Python scripts)
│   └── main.py
├── data/                # Input data or downloaded datasets
├── models/              # Saved machine learning models or checkpoints
├── requirements.txt     # List of Python dependencies
├── README.md            # Description of your project and instructions
├── .gitignore           # Files and folders to exclude from version control
└── venv/                # Virtual environment (should not be checked into Git)
```

You can create these directories manually using `mkdir` in your terminal or by using a file explorer.

**README.md** is the first thing others see when they visit your repository. Describe your project’s purpose, how to install dependencies, how to run the code and any usage examples.

**requirements.txt** lists all Python packages your project needs. You can create it manually or generate it with `pip freeze > requirements.txt` after installing your packages in the virtual environment.

**.gitignore** tells Git which files to exclude (for example, the virtual environment folder, compiled Python files and datasets). You can generate a suitable `.gitignore` for Python projects using templates from [gitignore.io](https://www.toptal.com/developers/gitignore).



## Version control with Git and GitHub

Using version control helps you track changes, collaborate with teammates and back up your code. Git is the most popular tool, and GitHub hosts repositories in the cloud.

### 1. Sign up and create a repository

1. Go to [GitHub](https://github.com/) and create a free account if you don’t have one.
2. Once logged in, click the **+** icon in the top‑right corner and choose **New repository**.
3. Enter a name and description for your repository. You can initialize it with a `README.md` and optionally choose a license. Click **Create repository**.

### 2. Clone your repository to your computer

In your new repository page, click the green **Code** button, copy the HTTPS clone URL, then run these commands in your terminal:

```bash
# Navigate to the directory where you want to store the project
cd ~/path/to/folder

# Clone the remote repository to your local machine
git clone https://github.com/your-username/your-repository.git

cd your-repository
```

This creates a local copy of the repository on your computer. You should see the files listed when you run `ls` or `dir`.

### 3. Add, commit and push changes

After making changes to files in your project, you need to stage and commit them:

```bash
# Check the status of your repository
git status

# Stage a modified file (e.g. README.md) for commit
git add README.md

# Commit your staged changes with a message describing what you did
git commit -m "Update README with installation instructions"
```

To upload your commits to GitHub, run:

```bash
git push
```

Git will ask for your GitHub username and password or personal access token the first time you push. After pushing, refresh your repository page on GitHub to see the updated files.

Using these basic commands (`git clone`, `git add`, `git commit`, `git push`) you can collaborate with others and keep your work history organized.

## Daytona – secure compute for agents

**Daytona** provides fast, scalable and stateful infrastructure for AI agents. It creates lightweight sandboxes in under 90 ms, allowing your agent to run code in an isolated environment. Each sandbox is separated from your own infrastructure so untrusted code executes safely. Daytona supports massive parallelization with real‑time output streaming.

Daytona exposes a programmatic API for process execution, file system operations and Git integration. This lets you control sandboxes from Python or TypeScript.

The code example below shows how to spin up a Python sandbox, run a script, execute a shell command and tear it down. Note that the package may already be installed in the hackathon environment. If not, uncomment the `pip install` line.


`````

# Install Daytona if needed
# !pip install daytona

from daytona import Daytona, CreateSandboxParams

# Create a client and start a Python sandbox
client = Daytona()
params = CreateSandboxParams(language="python")
sandbox = client.create(params)

# Run a Python snippet inside the sandbox
response = sandbox.process.code_run('print("Hello from Daytona!")')
if response.exit_code != 0:
    print(f"Error: {response.exit_code} {response.result}")
else:
    print(response.result)

# Execute an operating system command (e.g. listing files)
response = sandbox.process.exec('echo "This runs in an isolated shell"', cwd="/home/daytona", timeout=10)
if response.exit_code != 0:
    print(f"Error: {response.exit_code} {response.result}")
else:
    print(response.result)

# Clean up
client.remove(sandbox)
`````


## Nosana – decentralized GPU compute

**Nosana** is an open GPU cloud that provides on‑demand compute for AI and high‑performance workloads. It offers distributed access to GPUs worldwide so you can run jobs exactly when you need them. The platform is designed for intensive workloads – from training and fine‑tuning models to inference and rendering. Nosana’s compute is flexible and scalable: you can scale up for peak demand and scale down when you’re done, paying only for what you use. Jobs are cost‑efficient and run in isolated environments with security and privacy protections.

### Installing and configuring the Nosana CLI

Nosana provides a command‑line interface (CLI) for posting jobs to the network. You can install it using npm (or yarn, pnpm or bun). After installation, verify that the CLI is available:

```bash
# Install the CLI globally
npm install -g @nosana/cli

# Check the version
nosana --version
```

When you first run the CLI, a new Solana keypair is generated in `~/.nosana/.nosana_key.json`. This keypair is your wallet. To run jobs you need some **NOS** and **SOL** tokens in this wallet; you can acquire them from exchanges or via swaps. Use `nosana address` to view your wallet address.

### Posting your first job (Hello World)

Once your wallet has funds, pick a GPU market from the [Nosana Explorer](https://dashboard.nosana.com/). Markets list available GPUs, their prices and the number of hosts. Create a JSON job definition specifying the container image and command to run, then post it using the CLI:

```json
{
  "version": 1,
  "type": "job",
  "meta": {
    "name": "hello-world",
    "description": "Run echo in a container"
  },
  "global": {
    "resources": {
      "gpu": 0,
      "cpu": 1,
      "memory": 1,
      "storage": 1
    }
  },
  "ops": [
    {
      "id": "echo",
      "template": "docker",
      "image": "ubuntu:latest",
      "command": "echo Hello, Nosana!"
    }
  ]
}
```

To post the job:

```bash
nosana job post path/to/job-definition.json
```

After the job runs, you can retrieve the logs and results using `nosana job get <jobID>`.



## Running GPT‑OSS with Nosana (advanced)

OpenAI released **gpt-oss-20B** and **gpt-oss-120B** models under an open Apache 2.0 licence. These models support a large context window and offer competitive reasoning and tool-use performance. Because they are huge (the 20B-parameter model requires tens of gigabytes of memory), it’s impractical to run them on a laptop. Nosana’s decentralized GPUs provide an affordable way to deploy these models.

### Example job definition for GPT‑OSS 20B

Below is a simplified job definition that starts `ollama` with the `gpt-oss:20b` model. It assumes that the Docker image `nosana/ollama:latest` includes the necessary executables and that you have configured a market with sufficient VRAM.

```json
{
  "version": 1,
  "type": "job",
  "meta": {
    "name": "gpt-oss-20b",
    "description": "Run GPT‑OSS 20B with Ollama"
  },
  "global": {
    "resources": {
      "gpu": 1,
      "cpu": 4,
      "memory": 32,
      "storage": 10
    }
  },
  "ops": [
    {
      "id": "ollama",
      "template": "docker",
      "image": "nosana/ollama:latest",
      "command": "bash -c \"ollama pull gpt-oss:20b && ollama serve --model gpt-oss:20b\"",
      "port": 11434
    }
  ]
}
```

When the job runs successfully, Nosana provides a public URL where your model is hosted. You can then send HTTP requests to `http://<service-url>/api/generate` to interact with the model.

**Note:** These jobs require substantial VRAM and may take time to download the model on first run. Check your market’s VRAM size and costs before submitting.



## Building your hackathon project

Here is a conceptual workflow combining the tools and setup described above:

1. **Plan your project:** Identify a real‑world problem your agent will solve. Outline the data sources it needs and the actions it must perform.
2. **Set up your environment:** Create a project folder, install Python and set up a virtual environment. Initialize a Git repository and push it to GitHub so you can collaborate and track changes.
3. **Collect data:** Use Bright Data’s APIs or browsing infrastructure to gather real‑time information. Respect website terms of service and adjust payloads for your target domain.
4. **Process and reason:** Use Daytona sandboxes to safely run Python scripts or call compiled code without worrying about security. The isolated environment lets you iterate quickly.
5. **Offload heavy tasks:** When your agent needs a large model or GPU‑bound workload, create a Nosana job definition and submit it via the CLI. Interact with the model through the service URL returned by Nosana.
6. **Integrate and test:** Combine these components into an agentic workflow in your Python scripts under `src/`. Test end‑to‑end execution in your local environment.
7. **Document and demo:** Write clear instructions in your README.md and include screenshots or diagrams. Prepare a short demo showing your agent collecting data, reasoning and executing actions. Use the Jupyter notebook to illustrate how each piece works.

By following this guide, even participants new to coding can bootstrap a project and leverage powerful tools like Daytona, Bright Data and Nosana to build a sophisticated AI agent.
